```python
import datetime
from datetime import timedelta
import yfinance as yf

# Map sector to root and exchange suffix (WTI crude oil on NYMEX)
SECTOR_TO_ROOT = {
    "oil": ("CL", ".NYM"),
    "crude_oil": ("CL", ".NYM"),
}

MONTH_CODES = ['F', 'G', 'H', 'J', 'K', 'M', 'N', 'Q', 'U', 'V', 'X', 'Z']  # Jan..Dec

def generate_contracts(root: str, suffix: str, months_ahead: int = 24):
    today = datetime.date.today()
    year = today.year
    month = today.month
    out = []
    for i in range(months_ahead):
        m = (month + i - 1) % 12
        y = year + (month + i - 1) // 12
        month_code = MONTH_CODES[m]
        out.append(f"{root}{month_code}{str(y)[-2:]}{suffix}")
    return out

def has_recent_data(ticker: str, days: int = 10) -> bool:
    end = datetime.date.today() + timedelta(days=1)
    start = end - timedelta(days=days)
    df = yf.download(ticker, start=start.strftime("%Y-%m-%d"), end=end.strftime("%Y-%m-%d"), progress=False)
    return not df.empty

# Input
sector = "oil"

# Generate and validate
root, suffix = SECTOR_TO_ROOT[sector]
contracts = generate_contracts(root, suffix, months_ahead=24)

available = []
for t in contracts:
    try:
        if has_recent_data(t, days=10):
            available.append(t)
    except Exception as e:
        # Ignore tickers that error out
        pass

print(f"Total generated (24 months): {len(contracts)}")
print(f"Available on Yahoo (data in last 10 days): {len(available)}")
print(available)
```
