```python
#!/usr/bin/env python3
"""
Chain of Thought Agent - Generates linear impact chains with dynamic Mermaid.js
"""

import json
import re
from typing import Dict, Any, Optional, List
from pydantic import BaseModel, Field
from LLM_Call_Agent import LLMCallAgent

# ===== PYDANTIC MODEL =====
class ChainOfThoughtResult(BaseModel):
    """Structured output for chain of thought analysis"""
    initial_query: str = Field(description="The original user question")
    ticker: str = Field(description="Stock ticker symbol")
    impact_chain: str = Field(description="Linear chain: A → B → C → D → Final")
    final_direction: str = Field(description="Final impact: 'Short' or 'Long'")
    chain_explanation: str = Field(description="Brief explanation of the chain logic")
    node_count: int = Field(description="Number of nodes in the chain")
    edge_count: int = Field(description="Number of edges/connections in the chain")
    events: List[str] = Field(description="List of individual events in the chain")

# ===== TRULY DYNAMIC MERMAID.JS GENERATOR =====
def generate_mermaid_code(node_count: int, edge_count: int, events: List[str]) -> str:
    """
    ✅ TRULY DYNAMIC - Generates Mermaid.js code for ANY number of nodes
    """
    
    print(f"🔍 DEBUG: generate_mermaid_code called with:")
    print(f"   - node_count: {node_count}")
    print(f"   - edge_count: {edge_count}")
    print(f"   - events: {events}")
    
    # ✅ NO MINIMUM LIMITS - Let it be as short or long as needed
    if node_count < 1:  # Allow single node if needed
        node_count = 1
    if edge_count < 0:  # Allow no edges if single node
        edge_count = 0
    
    # ✅ DYNAMIC NODE IDS - Handle any number of nodes
    node_ids = []
    for i in range(node_count):
        if i < 26:  # A-Z
            node_ids.append(chr(65 + i))
        else:  # AA, AB, AC... for more than 26 nodes
            node_ids.append(f"{chr(65 + i//26)}{chr(65 + i%26)}")
    
    print(f"�� DEBUG: node_ids: {node_ids}")
    
    # ✅ CLEAN EVENTS - Remove any Pydantic field names
    clean_events = []
    for event in events[:node_count]:
        # Remove field names like "initial_query:", "ticker:", etc.
        if ': ' in event:
            clean_event = event.split(': ')[-1]
        else:
            clean_event = event
        clean_events.append(clean_event)
    
    print(f"🔍 DEBUG: clean_events: {clean_events}")
    
    # ✅ TRULY DYNAMIC MERMAID CODE - Adapts to any number of nodes
    mermaid_code = "graph LR\n"
    
    # ✅ FIXED - Create separate node for each event
    for i in range(node_count):
        node_id = node_ids[i]
        if i < len(clean_events):
            # ✅ LAST NODE - Just show the final decision (Short/Long)
            if i == node_count - 1:  # Last node
                # Extract just "Short" or "Long" from the last event
                last_event = clean_events[i]
                print(f"�� DEBUG: Last event: {last_event}")
                if "Short" in last_event:
                    event_text = "Short"
                elif "Long" in last_event:
                    event_text = "Long"
                else:
                    event_text = last_event
                print(f"🔍 DEBUG: Last node text: {event_text}")
            else:
                event_text = clean_events[i]
        else:
            event_text = f"Step {i+1}"
        mermaid_code += f"    {node_id}[{event_text}]\n"
    
    # Add edges dynamically
    for i in range(edge_count):
        if i + 1 < len(node_ids):
            source = node_ids[i]
            target = node_ids[i + 1]
            mermaid_code += f"    {source} --> {target}\n"
    
    # ✅ DYNAMIC STYLING - Adapts to any number of nodes
    if node_count > 0:
        mermaid_code += f"\n    style {node_ids[0]} fill:#e1f5fe\n"  # First node (blue)
        if node_count > 1:
            mermaid_code += f"    style {node_ids[-1]} fill:#ffebee\n"  # Last node (red)
            # Middle nodes (purple) - only if more than 2 nodes
            if node_count > 2:
                for i in range(1, node_count - 1):
                    node_id = node_ids[i]
                    mermaid_code += f"    style {node_id} fill:#f3e5f5\n"
    
    print(f"🔍 DEBUG: Final mermaid_code: {mermaid_code}")
    return mermaid_code
    
# ===== SMART RESPONSE PARSER =====
def parse_llm_response(response: str) -> Optional[Dict[str, Any]]:
    """
    ✅ TRULY DYNAMIC - Parses any chain length, not just 5 nodes
    """
    
    # Try to extract JSON first
    try:
        # Look for JSON-like content
        json_match = re.search(r'\{.*\}', response, re.DOTALL)
        if json_match:
            json_str = json_match.group(0)
            parsed = json.loads(json_str)
            return parsed
    except:
        pass
    
    # ✅ TRULY DYNAMIC - Look for ANY chain pattern, not just 5 nodes
    try:
        # Look for ANY chain pattern: A → B → C or A → B → C → D or A → B → C → D → E → F
        chain_match = re.search(r'([^→]+(?:→[^→]+)*)', response)
        if chain_match:
            chain_text = chain_match.group(1)
            chain_parts = [part.strip() for part in chain_text.split("→")]
            
            # ✅ DYNAMIC - Handle any number of parts
            if len(chain_parts) >= 2:  # At least 2 parts for a chain
                # Determine final direction
                final_direction = "Short" if "Short" in chain_parts[-1] else "Long"
                
                return {
                    "impact_chain": chain_text,
                    "final_direction": final_direction,
                    "node_count": len(chain_parts),  # ✅ DYNAMIC
                    "edge_count": len(chain_parts) - 1,  # ✅ DYNAMIC
                    "events": chain_parts
                }
    except:
        pass
    
    return None

# ===== CHAIN OF THOUGHT AGENT =====
class ChainOfThoughtAgent:
    def __init__(self):
        """Initialize the agent with centralized LLM"""
        self.llm_agent = LLMCallAgent(
            default_provider="deepseek",
            default_model="deepseek-chat"
        )
        
        # Get structured LLM for chain generation (without response_format)
        try:
            self.structured_llm = self.llm_agent.get_structured_llm(ChainOfThoughtResult)
        except Exception as e:
            print(f"⚠️ Could not create structured LLM: {e}")
            self.structured_llm = None
        
        print("🤖 Chain of Thought Agent initialized")
        print("🔑 API Keys Status:")
        print(f"   - DeepSeek: {'✅ Available' if self.llm_agent.deepseek_api_key else '❌ Missing'}")
        print(f"   - OpenAI: {'✅ Available' if self.llm_agent.openai_api_key else '❌ Missing'}")

    def generate_impact_chain(self, ticker: str, user_question: str, verification_links: List[str] = None, verification_reasoning: str = "", agent_analysis_results: Dict = None) -> ChainOfThoughtResult:
        """
        ✅ TRULY DYNAMIC - Generates chains of ANY length based on content
        """
        
        # Set default values if None
        if verification_links is None:
            verification_links = []
        if agent_analysis_results is None:
            agent_analysis_results = {}
        
        # ✅ TRULY DYNAMIC PROMPT - Explains causal chain and infers future impacts
        # ✅ TRULY DYNAMIC PROMPT - Focus on Answer part for inference support
        # ✅ TRULY DYNAMIC PROMPT - Data/Numerical/Evidence-Driven Events
        # ✅ TRULY DYNAMIC PROMPT - Short Last Node
        prompt = f"""
You are a Chain of Thought Agent. Your job is to create a DETAILED, EVIDENCE-BASED LINEAR IMPACT CHAIN showing how specific events CAUSALLY IMPACT each other and ultimately affect the asset's short-term future.

INPUT DATA:
- Ticker: {ticker}
- User Question: {user_question}
- Verification Links: {verification_links}  # ✅ Extra source for current asset context
- Verification Reasoning: {verification_reasoning}  # ✅ Extra source for current asset context
- Agent Analysis Results: {agent_analysis_results}  # ✅ Extra source for current asset context

TASK:
Create a DETAILED linear chain showing the SPECIFIC CAUSAL IMPACT FLOW. Each step must EXPLAIN HOW the previous event CAUSES the next event, leading to the final investment decision.

REQUIREMENTS:
1. ✅ CHAIN LENGTH IS DYNAMIC - Create as many steps as needed for the story (could be 2, 3, 4, 5, 6, or more)
2. ✅ EACH EVENT MUST BE DATA/NUMERICAL/EVIDENCE-DRIVEN:
   - Include ACTUAL NUMBERS (e.g., "Q2 EPS $0.12 vs $1.49 estimate")
   - Include SPECIFIC DATES (e.g., "July 31, 2025")
   - Include PERCENTAGES (e.g., "15% stock drop")
   - Include DOLLAR AMOUNTS (e.g., "$2B convertible notes")
   - Include ANALYST RATINGS (e.g., "Compass Point downgrade to Sell")
   - Include VOLUME DATA (e.g., "Trading volume declines 22%")
   - Include PRICE LEVELS (e.g., "Stock drops from $150 to $123")
3. ✅ EXPLAIN THE CAUSAL LINK - Show HOW Event A CAUSES Event B, Event B CAUSES Event C, etc.
4. ✅ INFER FUTURE IMPACTS - Project how current events will affect the asset in the short-term future
5. ✅ LAST NODE MUST BE SHORT - Just "Short" or "Long" (not long text)
6. ✅ FOCUS ON AGENT ANALYSIS RESULTS - Use the Answer part as your PRIMARY SOURCE for inference support points
7. ✅ USE VERIFICATION DATA AS CONTEXT - These provide extra context about the current asset state
8. ✅ NO TEMPLATE LOCKING - Let the story determine the length

EXAMPLE FORMATS (WITH SHORT LAST NODE):
- 3 nodes: "Q2 EPS $0.12 vs $1.49 Estimate → Stock drops 15% to $123 on July 31 → Short"
- 4 nodes: "Revenue $1.49B misses $1.52B estimate → Price declines 8% in 2 days → JPMorgan downgrades to Underweight → Short"
- 5 nodes: "Q2 EPS $0.12 vs $1.49 Estimate → 15% Stock Drop to $123 on July 31 → $2B Convertible Notes at 3.5% yield → Compass Point downgrade to Sell → Short"
- 6+ nodes: "Event A (with data) → Event B (with data, caused by A) → Event C (with data, caused by B) → Event D (with data, caused by C) → Event E (with data, caused by D) → Short"

OUTPUT FORMAT:
- initial_query: Copy the user question exactly
- ticker: The stock ticker
- impact_chain: Your DETAILED linear chain with DATA-DRIVEN events and causal explanations
- final_direction: "Short" or "Long"
- chain_explanation: Detailed explanation of the CAUSAL CHAIN and FUTURE PROJECTION based on the Answer part
- node_count: Number of nodes in your chain (DYNAMIC - could be any number)
- edge_count: Number of edges/connections (DYNAMIC - node_count - 1)
- events: List of DATA-DRIVEN events with causal explanations (DYNAMIC length)

IMPORTANT: 
- ✅ EVERY EVENT MUST CONTAIN SPECIFIC DATA/NUMBERS/DATES for precise inference
- ✅ LAST NODE MUST BE JUST "Short" or "Long" (not long text)
- ✅ Use the AGENT ANALYSIS RESULTS (Answer part) as your PRIMARY SOURCE for inference support points
- ✅ Use VERIFICATION DATA as extra context to understand the current asset state
- ✅ Explain HOW each event causes the next based on the evidence in the Answer part
- ✅ Project the FUTURE IMPACT on the asset using the inference support from the Answer part
- ✅ NO VAGUE EVENTS - Every step must have concrete, measurable data

Generate the DETAILED CAUSAL CHAIN now. Focus on the Answer part for inference support, ensure EVERY EVENT is data/numerical/evidence-driven, and make the LAST NODE just "Short" or "Long":
"""
        # Try structured LLM first
        if self.structured_llm:
            try:
                result = self.structured_llm.invoke(prompt)
                return result
            except Exception as e:
                print(f"❌ Structured LLM failed: {e}")
                print("🔄 Falling back to manual parsing...")
        
        # ✅ TRULY DYNAMIC FALLBACK - Use regular LLM and parse manually
        try:
            # Use the correct method from LLM_Call_Agent
            response = self.llm_agent.call_deepseek(
                prompt=prompt,
                system_message="You are a Chain of Thought Agent that creates linear impact chains for investment decisions. Create chains of ANY length based on the story.",
                max_tokens=1000,
                temperature=0.3
            )
            
            print(f"🤖 LLM Response: {response[:200]}...")
            
            # Use smart parser
            parsed_data = parse_llm_response(response)
            
            if parsed_data:
                # Extract chain parts
                if "impact_chain" in parsed_data:
                    chain_text = parsed_data["impact_chain"]
                    chain_parts = [part.strip() for part in chain_text.split("→")]
                    
                    # ✅ CLEAN EVENTS - Remove any Pydantic field artifacts
                    clean_events = []
                    for part in chain_parts:
                        # Remove field names and clean up
                        clean_part = part
                        if '- ' in clean_part:
                            clean_part = clean_part.split('- ')[-1]
                        if ': ' in clean_part:
                            clean_part = clean_part.split(': ')[-1]
                        clean_events.append(clean_part.strip())
                    
                    return ChainOfThoughtResult(
                        initial_query=user_question,
                        ticker=ticker,
                        impact_chain=chain_text,
                        final_direction=parsed_data.get("final_direction", "Short"),
                        chain_explanation=f"Generated from LLM response: {response[:200]}...",
                        node_count=len(clean_events),  # ✅ TRULY DYNAMIC
                        edge_count=len(clean_events) - 1,  # ✅ TRULY DYNAMIC
                        events=clean_events  # ✅ TRULY DYNAMIC
                    )
            
            # ✅ NO FALLBACK - If parsing fails, return error
            raise Exception("Failed to parse LLM response and no fallback available")
            
        except Exception as e:
            print(f"❌ All methods failed: {e}")
            # ✅ NO FALLBACK - Return error instead of preset
            raise Exception(f"Chain generation failed: {e}")

# ===== USAGE EXAMPLE (Jupyter Notebook Compatible) =====
def test_chain_agent():
    """Test the Chain of Thought Agent (no async needed)"""
    
    # Create agent and generate chain
    agent = ChainOfThoughtAgent()
    
    # Example parameters
    ticker = "COIN"
    user_question = "In late July, Coinbase reported Q2 2025 earnings that missed expectations:"
    verification_links = ["https://example.com/news1", "https://example.com/news2"]
    verification_reasoning = "Multiple credible sources confirm the earnings miss"
    agent_analysis_results = {"market_analysis": "Bearish sentiment due to earnings miss"}
    
    try:
        result = agent.generate_impact_chain(
            ticker=ticker,
            user_question=user_question,
            verification_links=verification_links,
            verification_reasoning=verification_reasoning,
            agent_analysis_results=agent_analysis_results
        )
        
        # Display results
        print("\n" + "=" * 60)
        print("🔗 CHAIN OF THOUGHT RESULT")
        print("=" * 60)
        print(f"Ticker: {result.ticker}")
        print(f"Query: {result.initial_query}")
        print(f"Impact Chain: {result.impact_chain}")
        print(f"Final Direction: {result.final_direction}")
        print(f"Explanation: {result.chain_explanation}")
        print(f"Node Count: {result.node_count}")
        print(f"Edge Count: {result.edge_count}")
        print(f"Events: {result.events}")
        
        # Generate dynamic Mermaid.js code
        mermaid_code = generate_mermaid_code(
            node_count=result.node_count,
            edge_count=result.edge_count,
            events=result.events
        )
        
        print(f"\n Dynamic Mermaid.js Code:")
        print(mermaid_code)
        
        return result, mermaid_code
        
    except Exception as e:
        print(f"❌ Chain generation failed: {e}")
        return None, None

```

```python
#!/usr/bin/env python3
"""
Chain of Thought Agent - Generates linear impact chains with dynamic Mermaid.js
"""

import json
import re
from typing import Dict, Any, Optional, List
from pydantic import BaseModel, Field
from LLM_Call_Agent import LLMCallAgent

# ===== PYDANTIC MODEL =====
class ChainOfThoughtResult(BaseModel):
    """Structured output for chain of thought analysis"""
    initial_query: str = Field(description="The original user question")
    ticker: str = Field(description="Stock ticker symbol")
    impact_chain: str = Field(description="Linear chain: A → B → C → D → Final")
    final_direction: str = Field(description="Final impact: 'Short' or 'Long'")
    chain_explanation: str = Field(description="Brief explanation of the chain logic")
    node_count: int = Field(description="Number of nodes in the chain")
    edge_count: int = Field(description="Number of edges/connections in the chain")
    events: List[str] = Field(description="List of individual events in the chain")

# ===== TRULY DYNAMIC MERMAID.JS GENERATOR =====
def generate_mermaid_code(node_count: int, edge_count: int, events: List[str]) -> str:
    """
    ✅ TRULY DYNAMIC - Generates Mermaid.js code for ANY number of nodes
    """
    
    print(f"🔍 DEBUG: generate_mermaid_code called with:")
    print(f"   - node_count: {node_count}")
    print(f"   - edge_count: {edge_count}")
    print(f"   - events: {events}")
    
    # ✅ NO MINIMUM LIMITS - Let it be as short or long as needed
    if node_count < 1:  # Allow single node if needed
        node_count = 1
    if edge_count < 0:  # Allow no edges if single node
        edge_count = 0
    
    # ✅ DYNAMIC NODE IDS - Handle any number of nodes
    node_ids = []
    for i in range(node_count):
        if i < 26:  # A-Z
            node_ids.append(chr(65 + i))
        else:  # AA, AB, AC... for more than 26 nodes
            node_ids.append(f"{chr(65 + i//26)}{chr(65 + i%26)}")
    
    print(f"�� DEBUG: node_ids: {node_ids}")
    
    # ✅ CLEAN EVENTS - Remove any Pydantic field names
    clean_events = []
    for event in events[:node_count]:
        # Remove field names like "initial_query:", "ticker:", etc.
        if ': ' in event:
            clean_event = event.split(': ')[-1]
        else:
            clean_event = event
        clean_events.append(clean_event)
    
    print(f"🔍 DEBUG: clean_events: {clean_events}")
    
    # ✅ TRULY DYNAMIC MERMAID CODE - Adapts to any number of nodes
    mermaid_code = "graph LR\n"
    
    # ✅ FIXED - Create separate node for each event
    for i in range(node_count):
        node_id = node_ids[i]
        if i < len(clean_events):
            # ✅ LAST NODE - Just show the final decision (Short/Long)
            if i == node_count - 1:  # Last node
                # Extract just "Short" or "Long" from the last event
                last_event = clean_events[i]
                print(f"�� DEBUG: Last event: {last_event}")
                if "Short" in last_event:
                    event_text = "Short"
                elif "Long" in last_event:
                    event_text = "Long"
                else:
                    event_text = last_event
                print(f"🔍 DEBUG: Last node text: {event_text}")
            else:
                event_text = clean_events[i]
        else:
            event_text = f"Step {i+1}"
        mermaid_code += f"    {node_id}[{event_text}]\n"
    
    # Add edges dynamically
    for i in range(edge_count):
        if i + 1 < len(node_ids):
            source = node_ids[i]
            target = node_ids[i + 1]
            mermaid_code += f"    {source} --> {target}\n"
    
    # ✅ DYNAMIC STYLING - Adapts to any number of nodes
    if node_count > 0:
        mermaid_code += f"\n    style {node_ids[0]} fill:#e1f5fe\n"  # First node (blue)
        if node_count > 1:
            mermaid_code += f"    style {node_ids[-1]} fill:#ffebee\n"  # Last node (red)
            # Middle nodes (purple) - only if more than 2 nodes
            if node_count > 2:
                for i in range(1, node_count - 1):
                    node_id = node_ids[i]
                    mermaid_code += f"    style {node_id} fill:#f3e5f5\n"
    
    print(f"🔍 DEBUG: Final mermaid_code: {mermaid_code}")
    return mermaid_code
    
# ===== SMART RESPONSE PARSER =====
def parse_llm_response(response: str) -> Optional[Dict[str, Any]]:
    """
    ✅ TRULY DYNAMIC - Parses any chain length, not just 5 nodes
    """
    
    # Try to extract JSON first
    try:
        # Look for JSON-like content
        json_match = re.search(r'\{.*\}', response, re.DOTALL)
        if json_match:
            json_str = json_match.group(0)
            parsed = json.loads(json_str)
            return parsed
    except:
        pass
    
    # ✅ TRULY DYNAMIC - Look for ANY chain pattern, not just 5 nodes
    try:
        # Look for ANY chain pattern: A → B → C or A → B → C → D or A → B → C → D → E → F
        chain_match = re.search(r'([^→]+(?:→[^→]+)*)', response)
        if chain_match:
            chain_text = chain_match.group(1)
            chain_parts = [part.strip() for part in chain_text.split("→")]
            
            # ✅ DYNAMIC - Handle any number of parts
            if len(chain_parts) >= 2:  # At least 2 parts for a chain
                # Determine final direction
                final_direction = "Short" if "Short" in chain_parts[-1] else "Long"
                
                return {
                    "impact_chain": chain_text,
                    "final_direction": final_direction,
                    "node_count": len(chain_parts),  # ✅ DYNAMIC
                    "edge_count": len(chain_parts) - 1,  # ✅ DYNAMIC
                    "events": chain_parts
                }
    except:
        pass
    
    return None

# ===== CHAIN OF THOUGHT AGENT =====
class ChainOfThoughtAgent:
    def __init__(self):
        """Initialize the agent with centralized LLM"""
        self.llm_agent = LLMCallAgent(
            default_provider="deepseek",
            default_model="deepseek-chat"
        )
        
        # Get structured LLM for chain generation (without response_format)
        try:
            self.structured_llm = self.llm_agent.get_structured_llm(ChainOfThoughtResult)
        except Exception as e:
            print(f"⚠️ Could not create structured LLM: {e}")
            self.structured_llm = None
        
        print("🤖 Chain of Thought Agent initialized")
        print("🔑 API Keys Status:")
        print(f"   - DeepSeek: {'✅ Available' if self.llm_agent.deepseek_api_key else '❌ Missing'}")
        print(f"   - OpenAI: {'✅ Available' if self.llm_agent.openai_api_key else '❌ Missing'}")

    def generate_impact_chain(self, ticker: str, user_question: str, verification_links: List[str] = None, verification_reasoning: str = "", agent_analysis_results: Dict = None) -> ChainOfThoughtResult:
        """
        ✅ TRULY DYNAMIC - Generates chains of ANY length based on content
        """
        
        # Set default values if None
        if verification_links is None:
            verification_links = []
        if agent_analysis_results is None:
            agent_analysis_results = {}
        
        # ✅ TRULY DYNAMIC PROMPT - Explains causal chain and infers future impacts
        # ✅ TRULY DYNAMIC PROMPT - Focus on Answer part for inference support
        # ✅ TRULY DYNAMIC PROMPT - Data/Numerical/Evidence-Driven Events
        # ✅ TRULY DYNAMIC PROMPT - Short Last Node
        prompt = f"""
You are a Chain of Thought Agent. Your job is to create a DETAILED, EVIDENCE-BASED LINEAR IMPACT CHAIN showing how specific events CAUSALLY IMPACT each other and ultimately affect the asset's short-term future.

INPUT DATA:
- Ticker: {ticker}
- User Question: {user_question}
- Verification Links: {verification_links}  # ✅ Extra source for current asset context
- Verification Reasoning: {verification_reasoning}  # ✅ Extra source for current asset context
- Agent Analysis Results: {agent_analysis_results}  # ✅ Extra source for current asset context

TASK:
Create a DETAILED linear chain showing the SPECIFIC CAUSAL IMPACT FLOW. Each step must EXPLAIN HOW the previous event CAUSES the next event, leading to the final investment decision.

REQUIREMENTS:
1. ✅ CHAIN LENGTH IS DYNAMIC - Create as many steps as needed for the story (could be 2, 3, 4, 5, 6, or more)
2. ✅ EACH EVENT MUST BE DATA/NUMERICAL/EVIDENCE-DRIVEN:
   - Include ACTUAL NUMBERS (e.g., "Q2 EPS $0.12 vs $1.49 estimate")
   - Include SPECIFIC DATES (e.g., "July 31, 2025")
   - Include PERCENTAGES (e.g., "15% stock drop")
   - Include DOLLAR AMOUNTS (e.g., "$2B convertible notes")
   - Include ANALYST RATINGS (e.g., "Compass Point downgrade to Sell")
   - Include VOLUME DATA (e.g., "Trading volume declines 22%")
   - Include PRICE LEVELS (e.g., "Stock drops from $150 to $123")
3. ✅ EXPLAIN THE CAUSAL LINK - Show HOW Event A CAUSES Event B, Event B CAUSES Event C, etc.
4. ✅ INFER FUTURE IMPACTS - Project how current events will affect the asset in the short-term future
5. ✅ LAST NODE MUST BE SHORT - Just "Short" or "Long" (not long text)
6. ✅ FOCUS ON AGENT ANALYSIS RESULTS - Use the Answer part as your PRIMARY SOURCE for inference support points
7. ✅ USE VERIFICATION DATA AS CONTEXT - These provide extra context about the current asset state
8. ✅ NO TEMPLATE LOCKING - Let the story determine the length

EXAMPLE FORMATS (WITH SHORT LAST NODE):
- 3 nodes: "Q2 EPS $0.12 vs $1.49 Estimate → Stock drops 15% to $123 on July 31 → Short"
- 4 nodes: "Revenue $1.49B misses $1.52B estimate → Price declines 8% in 2 days → JPMorgan downgrades to Underweight → Short"
- 5 nodes: "Q2 EPS $0.12 vs $1.49 Estimate → 15% Stock Drop to $123 on July 31 → $2B Convertible Notes at 3.5% yield → Compass Point downgrade to Sell → Short"
- 6+ nodes: "Event A (with data) → Event B (with data, caused by A) → Event C (with data, caused by B) → Event D (with data, caused by C) → Event E (with data, caused by D) → Short"

OUTPUT FORMAT:
- initial_query: Copy the user question exactly
- ticker: The stock ticker
- impact_chain: Your DETAILED linear chain with DATA-DRIVEN events and causal explanations
- final_direction: "Short" or "Long"
- chain_explanation: Detailed explanation of the CAUSAL CHAIN and FUTURE PROJECTION based on the Answer part
- node_count: Number of nodes in your chain (DYNAMIC - could be any number)
- edge_count: Number of edges/connections (DYNAMIC - node_count - 1)
- events: List of DATA-DRIVEN events with causal explanations (DYNAMIC length)

IMPORTANT: 
- ✅ EVERY EVENT MUST CONTAIN SPECIFIC DATA/NUMBERS/DATES for precise inference
- ✅ LAST NODE MUST BE JUST "Short" or "Long" (not long text)
- ✅ Use the AGENT ANALYSIS RESULTS (Answer part) as your PRIMARY SOURCE for inference support points
- ✅ Use VERIFICATION DATA as extra context to understand the current asset state
- ✅ Explain HOW each event causes the next based on the evidence in the Answer part
- ✅ Project the FUTURE IMPACT on the asset using the inference support from the Answer part
- ✅ NO VAGUE EVENTS - Every step must have concrete, measurable data

Generate the DETAILED CAUSAL CHAIN now. Focus on the Answer part for inference support, ensure EVERY EVENT is data/numerical/evidence-driven, and make the LAST NODE just "Short" or "Long":
"""
        # Try structured LLM first
        if self.structured_llm:
            try:
                result = self.structured_llm.invoke(prompt)
                return result
            except Exception as e:
                print(f"❌ Structured LLM failed: {e}")
                print("🔄 Falling back to manual parsing...")
        
        # ✅ TRULY DYNAMIC FALLBACK - Use regular LLM and parse manually
        try:
            # Use the correct method from LLM_Call_Agent
            response = self.llm_agent.call_deepseek(
                prompt=prompt,
                system_message="You are a Chain of Thought Agent that creates linear impact chains for investment decisions. Create chains of ANY length based on the story.",
                max_tokens=1000,
                temperature=0.3
            )
            
            print(f"🤖 LLM Response: {response[:200]}...")
            
            # Use smart parser
            parsed_data = parse_llm_response(response)
            
            if parsed_data:
                # Extract chain parts
                if "impact_chain" in parsed_data:
                    chain_text = parsed_data["impact_chain"]
                    chain_parts = [part.strip() for part in chain_text.split("→")]
                    
                    # ✅ CLEAN EVENTS - Remove any Pydantic field artifacts
                    clean_events = []
                    for part in chain_parts:
                        # Remove field names and clean up
                        clean_part = part
                        if '- ' in clean_part:
                            clean_part = clean_part.split('- ')[-1]
                        if ': ' in clean_part:
                            clean_part = clean_part.split(': ')[-1]
                        clean_events.append(clean_part.strip())
                    
                    return ChainOfThoughtResult(
                        initial_query=user_question,
                        ticker=ticker,
                        impact_chain=chain_text,
                        final_direction=parsed_data.get("final_direction", "Short"),
                        chain_explanation=f"Generated from LLM response: {response[:200]}...",
                        node_count=len(clean_events),  # ✅ TRULY DYNAMIC
                        edge_count=len(clean_events) - 1,  # ✅ TRULY DYNAMIC
                        events=clean_events  # ✅ TRULY DYNAMIC
                    )
            
            # ✅ NO FALLBACK - If parsing fails, return error
            raise Exception("Failed to parse LLM response and no fallback available")
            
        except Exception as e:
            print(f"❌ All methods failed: {e}")
            # ✅ NO FALLBACK - Return error instead of preset
            raise Exception(f"Chain generation failed: {e}")

# ===== USAGE EXAMPLE (Jupyter Notebook Compatible) =====
def test_chain_agent():
    """Test the Chain of Thought Agent (no async needed)"""
    
    # Create agent and generate chain
    agent = ChainOfThoughtAgent()
    
    # Example parameters
    ticker = "COIN"
    user_question = "In late July, Coinbase reported Q2 2025 earnings that missed expectations:"
    verification_links = ["https://example.com/news1", "https://example.com/news2"]
    verification_reasoning = "Multiple credible sources confirm the earnings miss"
    agent_analysis_results = {"market_analysis": "Bearish sentiment due to earnings miss"}
    
    try:
        result = agent.generate_impact_chain(
            ticker=ticker,
            user_question=user_question,
            verification_links=verification_links,
            verification_reasoning=verification_reasoning,
            agent_analysis_results=agent_analysis_results
        )
        
        # Display results
        print("\n" + "=" * 60)
        print("🔗 CHAIN OF THOUGHT RESULT")
        print("=" * 60)
        print(f"Ticker: {result.ticker}")
        print(f"Query: {result.initial_query}")
        print(f"Impact Chain: {result.impact_chain}")
        print(f"Final Direction: {result.final_direction}")
        print(f"Explanation: {result.chain_explanation}")
        print(f"Node Count: {result.node_count}")
        print(f"Edge Count: {result.edge_count}")
        print(f"Events: {result.events}")
        
        # Generate dynamic Mermaid.js code
        mermaid_code = generate_mermaid_code(
            node_count=result.node_count,
            edge_count=result.edge_count,
            events=result.events
        )
        
        print(f"\n Dynamic Mermaid.js Code:")
        print(mermaid_code)
        
        return result, mermaid_code
        
    except Exception as e:
        print(f"❌ Chain generation failed: {e}")
        return None, None

```

```python
#!/usr/bin/env python3
"""
Chain of Thought Agent - Generates linear impact chains with dynamic Mermaid.js
"""

import json
import re
from typing import Dict, Any, Optional, List
from pydantic import BaseModel, Field
from LLM_Call_Agent import LLMCallAgent

# ===== PYDANTIC MODEL =====
class ChainOfThoughtResult(BaseModel):
    """Structured output for chain of thought analysis"""
    initial_query: str = Field(description="The original user question")
    ticker: str = Field(description="Stock ticker symbol")
    impact_chain: str = Field(description="Linear chain: A → B → C → D → Final")
    final_direction: str = Field(description="Final impact: 'Short' or 'Long'")
    chain_explanation: str = Field(description="Brief explanation of the chain logic")
    node_count: int = Field(description="Number of nodes in the chain")
    edge_count: int = Field(description="Number of edges/connections in the chain")
    events: List[str] = Field(description="List of individual events in the chain")

# ===== TRULY DYNAMIC MERMAID.JS GENERATOR =====
def generate_mermaid_code(node_count: int, edge_count: int, events: List[str]) -> str:
    """
    ✅ TRULY DYNAMIC - Generates Mermaid.js code for ANY number of nodes
    """
    
    print(f"🔍 DEBUG: generate_mermaid_code called with:")
    print(f"   - node_count: {node_count}")
    print(f"   - edge_count: {edge_count}")
    print(f"   - events: {events}")
    
    # ✅ NO MINIMUM LIMITS - Let it be as short or long as needed
    if node_count < 1:  # Allow single node if needed
        node_count = 1
    if edge_count < 0:  # Allow no edges if single node
        edge_count = 0
    
    # ✅ DYNAMIC NODE IDS - Handle any number of nodes
    node_ids = []
    for i in range(node_count):
        if i < 26:  # A-Z
            node_ids.append(chr(65 + i))
        else:  # AA, AB, AC... for more than 26 nodes
            node_ids.append(f"{chr(65 + i//26)}{chr(65 + i%26)}")
    
    print(f"�� DEBUG: node_ids: {node_ids}")
    
    # ✅ CLEAN EVENTS - Remove any Pydantic field names
    clean_events = []
    for event in events[:node_count]:
        # Remove field names like "initial_query:", "ticker:", etc.
        if ': ' in event:
            clean_event = event.split(': ')[-1]
        else:
            clean_event = event
        clean_events.append(clean_event)
    
    print(f"🔍 DEBUG: clean_events: {clean_events}")
    
    # ✅ TRULY DYNAMIC MERMAID CODE - Adapts to any number of nodes
    mermaid_code = "graph LR\n"
    
    # ✅ FIXED - Create separate node for each event
    for i in range(node_count):
        node_id = node_ids[i]
        if i < len(clean_events):
            # ✅ LAST NODE - Just show the final decision (Short/Long)
            if i == node_count - 1:  # Last node
                # Extract just "Short" or "Long" from the last event
                last_event = clean_events[i]
                print(f"�� DEBUG: Last event: {last_event}")
                if "Short" in last_event:
                    event_text = "Short"
                elif "Long" in last_event:
                    event_text = "Long"
                else:
                    event_text = last_event
                print(f"🔍 DEBUG: Last node text: {event_text}")
            else:
                event_text = clean_events[i]
        else:
            event_text = f"Step {i+1}"
        mermaid_code += f"    {node_id}[{event_text}]\n"
    
    # Add edges dynamically
    for i in range(edge_count):
        if i + 1 < len(node_ids):
            source = node_ids[i]
            target = node_ids[i + 1]
            mermaid_code += f"    {source} --> {target}\n"
    
    # ✅ DYNAMIC STYLING - Adapts to any number of nodes
    if node_count > 0:
        mermaid_code += f"\n    style {node_ids[0]} fill:#e1f5fe\n"  # First node (blue)
        if node_count > 1:
            mermaid_code += f"    style {node_ids[-1]} fill:#ffebee\n"  # Last node (red)
            # Middle nodes (purple) - only if more than 2 nodes
            if node_count > 2:
                for i in range(1, node_count - 1):
                    node_id = node_ids[i]
                    mermaid_code += f"    style {node_id} fill:#f3e5f5\n"
    
    print(f"🔍 DEBUG: Final mermaid_code: {mermaid_code}")
    return mermaid_code
    
# ===== SMART RESPONSE PARSER =====
def parse_llm_response(response: str) -> Optional[Dict[str, Any]]:
    """
    ✅ TRULY DYNAMIC - Parses any chain length, not just 5 nodes
    """
    
    # Try to extract JSON first
    try:
        # Look for JSON-like content
        json_match = re.search(r'\{.*\}', response, re.DOTALL)
        if json_match:
            json_str = json_match.group(0)
            parsed = json.loads(json_str)
            return parsed
    except:
        pass
    
    # ✅ TRULY DYNAMIC - Look for ANY chain pattern, not just 5 nodes
    try:
        # Look for ANY chain pattern: A → B → C or A → B → C → D or A → B → C → D → E → F
        chain_match = re.search(r'([^→]+(?:→[^→]+)*)', response)
        if chain_match:
            chain_text = chain_match.group(1)
            chain_parts = [part.strip() for part in chain_text.split("→")]
            
            # ✅ DYNAMIC - Handle any number of parts
            if len(chain_parts) >= 2:  # At least 2 parts for a chain
                # Determine final direction
                final_direction = "Short" if "Short" in chain_parts[-1] else "Long"
                
                return {
                    "impact_chain": chain_text,
                    "final_direction": final_direction,
                    "node_count": len(chain_parts),  # ✅ DYNAMIC
                    "edge_count": len(chain_parts) - 1,  # ✅ DYNAMIC
                    "events": chain_parts
                }
    except:
        pass
    
    return None

# ===== CHAIN OF THOUGHT AGENT =====
class ChainOfThoughtAgent:
    def __init__(self):
        """Initialize the agent with centralized LLM"""
        self.llm_agent = LLMCallAgent(
            default_provider="deepseek",
            default_model="deepseek-chat"
        )
        
        # Get structured LLM for chain generation (without response_format)
        try:
            self.structured_llm = self.llm_agent.get_structured_llm(ChainOfThoughtResult)
        except Exception as e:
            print(f"⚠️ Could not create structured LLM: {e}")
            self.structured_llm = None
        
        print("🤖 Chain of Thought Agent initialized")
        print("🔑 API Keys Status:")
        print(f"   - DeepSeek: {'✅ Available' if self.llm_agent.deepseek_api_key else '❌ Missing'}")
        print(f"   - OpenAI: {'✅ Available' if self.llm_agent.openai_api_key else '❌ Missing'}")

    def generate_impact_chain(self, ticker: str, user_question: str, verification_links: List[str] = None, verification_reasoning: str = "", agent_analysis_results: Dict = None) -> ChainOfThoughtResult:
        """
        ✅ TRULY DYNAMIC - Generates chains of ANY length based on content
        """
        
        # Set default values if None
        if verification_links is None:
            verification_links = []
        if agent_analysis_results is None:
            agent_analysis_results = {}
        
        # ✅ TRULY DYNAMIC PROMPT - Explains causal chain and infers future impacts
        # ✅ TRULY DYNAMIC PROMPT - Focus on Answer part for inference support
        # ✅ TRULY DYNAMIC PROMPT - Data/Numerical/Evidence-Driven Events
        # ✅ TRULY DYNAMIC PROMPT - Short Last Node
        prompt = f"""
You are a Chain of Thought Agent. Your job is to create a DETAILED, EVIDENCE-BASED LINEAR IMPACT CHAIN showing how specific events CAUSALLY IMPACT each other and ultimately affect the asset's short-term future.

INPUT DATA:
- Ticker: {ticker}
- User Question: {user_question}
- Verification Links: {verification_links}  # ✅ Extra source for current asset context
- Verification Reasoning: {verification_reasoning}  # ✅ Extra source for current asset context
- Agent Analysis Results: {agent_analysis_results}  # ✅ Extra source for current asset context

TASK:
Create a DETAILED linear chain showing the SPECIFIC CAUSAL IMPACT FLOW. Each step must EXPLAIN HOW the previous event CAUSES the next event, leading to the final investment decision.

REQUIREMENTS:
1. ✅ CHAIN LENGTH IS DYNAMIC - Create as many steps as needed for the story (could be 2, 3, 4, 5, 6, or more)
2. ✅ EACH EVENT MUST BE DATA/NUMERICAL/EVIDENCE-DRIVEN:
   - Include ACTUAL NUMBERS (e.g., "Q2 EPS $0.12 vs $1.49 estimate")
   - Include SPECIFIC DATES (e.g., "July 31, 2025")
   - Include PERCENTAGES (e.g., "15% stock drop")
   - Include DOLLAR AMOUNTS (e.g., "$2B convertible notes")
   - Include ANALYST RATINGS (e.g., "Compass Point downgrade to Sell")
   - Include VOLUME DATA (e.g., "Trading volume declines 22%")
   - Include PRICE LEVELS (e.g., "Stock drops from $150 to $123")
3. ✅ EXPLAIN THE CAUSAL LINK - Show HOW Event A CAUSES Event B, Event B CAUSES Event C, etc.
4. ✅ INFER FUTURE IMPACTS - Project how current events will affect the asset in the short-term future
5. ✅ LAST NODE MUST BE SHORT - Just "Short" or "Long" (not long text)
6. ✅ FOCUS ON AGENT ANALYSIS RESULTS - Use the Answer part as your PRIMARY SOURCE for inference support points
7. ✅ USE VERIFICATION DATA AS CONTEXT - These provide extra context about the current asset state
8. ✅ NO TEMPLATE LOCKING - Let the story determine the length

EXAMPLE FORMATS (WITH SHORT LAST NODE):
- 3 nodes: "Q2 EPS $0.12 vs $1.49 Estimate → Stock drops 15% to $123 on July 31 → Short"
- 4 nodes: "Revenue $1.49B misses $1.52B estimate → Price declines 8% in 2 days → JPMorgan downgrades to Underweight → Short"
- 5 nodes: "Q2 EPS $0.12 vs $1.49 Estimate → 15% Stock Drop to $123 on July 31 → $2B Convertible Notes at 3.5% yield → Compass Point downgrade to Sell → Short"
- 6+ nodes: "Event A (with data) → Event B (with data, caused by A) → Event C (with data, caused by B) → Event D (with data, caused by C) → Event E (with data, caused by D) → Short"

OUTPUT FORMAT:
- initial_query: Copy the user question exactly
- ticker: The stock ticker
- impact_chain: Your DETAILED linear chain with DATA-DRIVEN events and causal explanations
- final_direction: "Short" or "Long"
- chain_explanation: Detailed explanation of the CAUSAL CHAIN and FUTURE PROJECTION based on the Answer part
- node_count: Number of nodes in your chain (DYNAMIC - could be any number)
- edge_count: Number of edges/connections (DYNAMIC - node_count - 1)
- events: List of DATA-DRIVEN events with causal explanations (DYNAMIC length)

IMPORTANT: 
- ✅ EVERY EVENT MUST CONTAIN SPECIFIC DATA/NUMBERS/DATES for precise inference
- ✅ LAST NODE MUST BE JUST "Short" or "Long" (not long text)
- ✅ Use the AGENT ANALYSIS RESULTS (Answer part) as your PRIMARY SOURCE for inference support points
- ✅ Use VERIFICATION DATA as extra context to understand the current asset state
- ✅ Explain HOW each event causes the next based on the evidence in the Answer part
- ✅ Project the FUTURE IMPACT on the asset using the inference support from the Answer part
- ✅ NO VAGUE EVENTS - Every step must have concrete, measurable data

Generate the DETAILED CAUSAL CHAIN now. Focus on the Answer part for inference support, ensure EVERY EVENT is data/numerical/evidence-driven, and make the LAST NODE just "Short" or "Long":
"""
        # Try structured LLM first
        if self.structured_llm:
            try:
                result = self.structured_llm.invoke(prompt)
                return result
            except Exception as e:
                print(f"❌ Structured LLM failed: {e}")
                print("🔄 Falling back to manual parsing...")
        
        # ✅ TRULY DYNAMIC FALLBACK - Use regular LLM and parse manually
        try:
            # Use the correct method from LLM_Call_Agent
            response = self.llm_agent.call_deepseek(
                prompt=prompt,
                system_message="You are a Chain of Thought Agent that creates linear impact chains for investment decisions. Create chains of ANY length based on the story.",
                max_tokens=1000,
                temperature=0.3
            )
            
            print(f"🤖 LLM Response: {response[:200]}...")
            
            # Use smart parser
            parsed_data = parse_llm_response(response)
            
            if parsed_data:
                # Extract chain parts
                if "impact_chain" in parsed_data:
                    chain_text = parsed_data["impact_chain"]
                    chain_parts = [part.strip() for part in chain_text.split("→")]
                    
                    # ✅ CLEAN EVENTS - Remove any Pydantic field artifacts
                    clean_events = []
                    for part in chain_parts:
                        # Remove field names and clean up
                        clean_part = part
                        if '- ' in clean_part:
                            clean_part = clean_part.split('- ')[-1]
                        if ': ' in clean_part:
                            clean_part = clean_part.split(': ')[-1]
                        clean_events.append(clean_part.strip())
                    
                    return ChainOfThoughtResult(
                        initial_query=user_question,
                        ticker=ticker,
                        impact_chain=chain_text,
                        final_direction=parsed_data.get("final_direction", "Short"),
                        chain_explanation=f"Generated from LLM response: {response[:200]}...",
                        node_count=len(clean_events),  # ✅ TRULY DYNAMIC
                        edge_count=len(clean_events) - 1,  # ✅ TRULY DYNAMIC
                        events=clean_events  # ✅ TRULY DYNAMIC
                    )
            
            # ✅ NO FALLBACK - If parsing fails, return error
            raise Exception("Failed to parse LLM response and no fallback available")
            
        except Exception as e:
            print(f"❌ All methods failed: {e}")
            # ✅ NO FALLBACK - Return error instead of preset
            raise Exception(f"Chain generation failed: {e}")

# ===== USAGE EXAMPLE (Jupyter Notebook Compatible) =====
def test_chain_agent():
    """Test the Chain of Thought Agent (no async needed)"""
    
    # Create agent and generate chain
    agent = ChainOfThoughtAgent()
    
    # Example parameters
    ticker = "COIN"
    user_question = "In late July, Coinbase reported Q2 2025 earnings that missed expectations:"
    verification_links = ["https://example.com/news1", "https://example.com/news2"]
    verification_reasoning = "Multiple credible sources confirm the earnings miss"
    agent_analysis_results = {"market_analysis": "Bearish sentiment due to earnings miss"}
    
    try:
        result = agent.generate_impact_chain(
            ticker=ticker,
            user_question=user_question,
            verification_links=verification_links,
            verification_reasoning=verification_reasoning,
            agent_analysis_results=agent_analysis_results
        )
        
        # Display results
        print("\n" + "=" * 60)
        print("🔗 CHAIN OF THOUGHT RESULT")
        print("=" * 60)
        print(f"Ticker: {result.ticker}")
        print(f"Query: {result.initial_query}")
        print(f"Impact Chain: {result.impact_chain}")
        print(f"Final Direction: {result.final_direction}")
        print(f"Explanation: {result.chain_explanation}")
        print(f"Node Count: {result.node_count}")
        print(f"Edge Count: {result.edge_count}")
        print(f"Events: {result.events}")
        
        # Generate dynamic Mermaid.js code
        mermaid_code = generate_mermaid_code(
            node_count=result.node_count,
            edge_count=result.edge_count,
            events=result.events
        )
        
        print(f"\n Dynamic Mermaid.js Code:")
        print(mermaid_code)
        
        return result, mermaid_code
        
    except Exception as e:
        print(f"❌ Chain generation failed: {e}")
        return None, None

```
